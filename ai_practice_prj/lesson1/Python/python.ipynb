### Trò chơi đoán từ


```python
# # trò  chơi đoán từ 
secret_word = "python"
hint=" gợi ý đây là 1  ngôn ngữ lập trình giống với 1 con vật dài thòn khôn chân"
guess=""
guess_count = 0
guess_limit = 3
print(hint)

while guess!=secret_word:
    if guess_count < guess_limit:
        guess = input("Bạn đoán từ gì? ")
        guess_count +=1
        if guess != secret_word and guess_count < guess_limit:
            print("nhap lai !")
    else:
        break
    
if guess == secret_word:
    print("Chính xác ! ")
else: 
    print("Thua rồi !")
```

### Máy tính lũy thừa

```python
#máy tính tính lũy thừa
#exponent là số mũ
def calculate_power(number,exponent):
    result = 1
    for i in range(exponent):
        result *= number
    return result

print("Nhap number va so mũ:")
so =int(input("nhap so number: "))
mu = int(input("nhap so mũ: "))
print(calculate_power(so,mu))
```

### Vòng lặp for
phân tích:
    -readlines() sẽ đọc tất cả các dòng trong file , mỗi dòng đọc ra sẽ cách nhau 1 dòng \n , nên sẽ dùng strip() để xóa nó đi

```python
#for
phone_book = open("phone_book.txt", "r", encoding="utf-8")
read_lines = phone_book.readlines()
phone_book.close()

# Cách sửa: Dùng range(len(read_lines))
for i in range(len(read_lines)):
    print(read_lines[i].strip())
```
